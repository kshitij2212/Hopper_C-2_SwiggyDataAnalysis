# 04 Statistical Analysis

This notebook performs statistical analysis on the cleaned Swiggy dataset:
1. Descriptive statistics grouped by City and Category
2. Average Price by Category and City
3. Hypothesis testing — ANOVA (Rating by City, Rating by Category)
4. Spearman correlation — Price vs Rating

All results are saved to `reports/stats/`.

```python
from pathlib import Path
import pandas as pd
from scipy import stats

current_dir = Path.cwd().resolve()
PROJECT_ROOT = current_dir.parent if current_dir.name.strip() == 'notebooks' else current_dir
STATS_PATH = PROJECT_ROOT / 'reports/stats'
STATS_PATH.mkdir(parents=True, exist_ok=True)

ALPHA = 0.05
```

```python
PROCESSED_PATH = PROJECT_ROOT / 'data/processed/swiggy_cleaned.csv'
df = pd.read_csv(PROCESSED_PATH, parse_dates=['order_date'])
print('Loaded:', df.shape)
df.head()
```

## 1. Descriptive Statistics — Grouped by City

```python
grouped_city = (
    df.groupby('city')[['price_inr', 'rating', 'rating_count']]
    .agg(['mean', 'median', 'std', 'min', 'max'])
    .round(2)
)
grouped_city.columns = ['_'.join(col) for col in grouped_city.columns]
grouped_city = grouped_city.reset_index()
grouped_city.to_csv(STATS_PATH / 'grouped_stats_by_city.csv', index=False)
print('Grouped stats by City saved.')
grouped_city
```

## 2. Descriptive Statistics — Grouped by Category

```python
grouped_category = (
    df.groupby('category')[['price_inr', 'rating', 'rating_count']]
    .agg(['mean', 'median', 'std', 'min', 'max'])
    .round(2)
)
grouped_category.columns = ['_'.join(col) for col in grouped_category.columns]
grouped_category = grouped_category.reset_index()
grouped_category.to_csv(STATS_PATH / 'grouped_stats_by_category.csv', index=False)
print('Grouped stats by Category saved.')
grouped_category
```

## 3. Average Price by Category

```python
avg_price_category = (
    df.groupby('category')['price_inr']
    .mean()
    .round(2)
    .reset_index()
    .rename(columns={'price_inr': 'avg_price_inr'})
    .sort_values('avg_price_inr', ascending=False)
)
avg_price_category.to_csv(STATS_PATH / 'avg_price_by_category.csv', index=False)
print('Average price by Category saved.')
avg_price_category
```

## 4. Average Price by City

```python
avg_price_city = (
    df.groupby('city')['price_inr']
    .mean()
    .round(2)
    .reset_index()
    .rename(columns={'price_inr': 'avg_price_inr'})
    .sort_values('avg_price_inr', ascending=False)
)
avg_price_city.to_csv(STATS_PATH / 'avg_price_by_city.csv', index=False)
print('Average price by City saved.')
avg_price_city
```

## 5. Hypothesis Testing

### 5a. ANOVA — Does Rating differ significantly across Cities?

```python
city_groups = [group['rating'].dropna().values for _, group in df.groupby('city')]
f_stat_city, p_val_city = stats.f_oneway(*city_groups)

print('ANOVA — Rating by City')
print(f'  F-statistic : {f_stat_city:.4f}')
print(f'  p-value     : {p_val_city:.4f}')
print(f'  Result      : {"Significant difference" if p_val_city < ALPHA else "No significant difference"} at α = {ALPHA}')
```

### 5b. ANOVA — Does Rating differ significantly across Categories?

```python
category_groups = [group['rating'].dropna().values for _, group in df.groupby('category')]
f_stat_cat, p_val_cat = stats.f_oneway(*category_groups)

print('ANOVA — Rating by Category')
print(f'  F-statistic : {f_stat_cat:.4f}')
print(f'  p-value     : {p_val_cat:.4f}')
print(f'  Result      : {"Significant difference" if p_val_cat < ALPHA else "No significant difference"} at α = {ALPHA}')
```

### 5c. Spearman Correlation — Price vs Rating

```python
corr_data = df[['price_inr', 'rating']].dropna()
spearman_corr, spearman_p = stats.spearmanr(corr_data['price_inr'], corr_data['rating'])

print('Spearman Correlation — Price vs Rating')
print(f'  Correlation : {spearman_corr:.4f}')
print(f'  p-value     : {spearman_p:.4f}')
print(f'  Result      : {"Significant correlation" if spearman_p < ALPHA else "No significant correlation"} at α = {ALPHA}')
```

## 6. Save Hypothesis Test Results

```python
hypothesis_results = pd.DataFrame([
    {
        'Test'        : 'ANOVA',
        'Description' : 'Rating by City',
        'F-Statistic' : round(f_stat_city, 4),
        'p-value'     : round(p_val_city, 4),
        'Significant' : p_val_city < ALPHA
    },
    {
        'Test'        : 'ANOVA',
        'Description' : 'Rating by Category',
        'F-Statistic' : round(f_stat_cat, 4),
        'p-value'     : round(p_val_cat, 4),
        'Significant' : p_val_cat < ALPHA
    },
    {
        'Test'        : 'Spearman Correlation',
        'Description' : 'Price vs Rating',
        'F-Statistic' : round(spearman_corr, 4),
        'p-value'     : round(spearman_p, 4),
        'Significant' : spearman_p < ALPHA
    }
])
hypothesis_results.to_csv(STATS_PATH / 'hypothesis_test_results.csv', index=False)
print('Hypothesis test results saved.')
hypothesis_results
```

## 7. Statistical Visualizations

Visual representations of the hypothesis testing results and correlation analysis.

```python
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import seaborn as sns

FIGURES_PATH = PROJECT_ROOT / 'reports/figures'
FIGURES_PATH.mkdir(parents=True, exist_ok=True)

sns.set_theme(style='whitegrid')
plt.rcParams['figure.dpi'] = 150

# Price data excluding outliers (same filter as EDA notebook)
df_clean_price = df[~df['is_price_outlier']].copy()
print('Rows used for price plots (no outliers):', len(df_clean_price))
```

### 7a. Box Plots — Price and Rating by City

```python
# Sort cities by median rating for the rating subplot
city_order_rating = (
    df.groupby('city')['rating']
    .median()
    .sort_values(ascending=True)
    .index.tolist()
)

# Sort cities by median price for the price subplot
city_order_price = (
    df_clean_price.groupby('city')['price_inr']
    .median()
    .sort_values(ascending=True)
    .index.tolist()
)

fig, axes = plt.subplots(1, 2, figsize=(18, 9))

# Price by City
sns.boxplot(
    data=df_clean_price, x='price_inr', y='city',
    order=city_order_price, palette='Blues_d', ax=axes[0]
)
axes[0].set_title('Price Distribution by City\n(outliers excluded)', fontsize=13, fontweight='bold')
axes[0].set_xlabel('Price (₹)', fontsize=11)
axes[0].set_ylabel('City', fontsize=11)
axes[0].xaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'₹{int(x)}'))

# Rating by City
sns.boxplot(
    data=df, x='rating', y='city',
    order=city_order_rating, palette='Greens_d', ax=axes[1]
)
axes[1].set_title('Rating Distribution by City\n(ANOVA: F=62.97, p<0.001)', fontsize=13, fontweight='bold')
axes[1].set_xlabel('Rating (1.5–5.0)', fontsize=11)
axes[1].set_ylabel('')

plt.suptitle('City-level Price & Rating Spread', fontsize=15, fontweight='bold', y=1.01)
plt.tight_layout()
plt.savefig(FIGURES_PATH / '11_boxplot_city.png', bbox_inches='tight', dpi=150)
plt.show()
print('Saved: 11_boxplot_city.png')
```

### 7b. Box Plots — Price and Rating by Top 10 Categories

```python
top10_cats = (
    df['category'].value_counts().head(10).index.tolist()
)
df_top10 = df[df['category'].isin(top10_cats)]
df_top10_clean = df_clean_price[df_clean_price['category'].isin(top10_cats)]

cat_order_price = (
    df_top10_clean.groupby('category')['price_inr']
    .median()
    .sort_values(ascending=True)
    .index.tolist()
)
cat_order_rating = (
    df_top10.groupby('category')['rating']
    .median()
    .sort_values(ascending=True)
    .index.tolist()
)

fig, axes = plt.subplots(1, 2, figsize=(20, 8))

sns.boxplot(
    data=df_top10_clean, x='price_inr', y='category',
    order=cat_order_price, palette='Oranges_d', ax=axes[0]
)
axes[0].set_title('Price Distribution by Category\n(Top 10, outliers excluded)', fontsize=13, fontweight='bold')
axes[0].set_xlabel('Price (₹)', fontsize=11)
axes[0].set_ylabel('Category', fontsize=11)
axes[0].xaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'₹{int(x)}'))

sns.boxplot(
    data=df_top10, x='rating', y='category',
    order=cat_order_rating, palette='Purples_d', ax=axes[1]
)
axes[1].set_title('Rating Distribution by Category\n(ANOVA: F=3.12, p<0.001)', fontsize=13, fontweight='bold')
axes[1].set_xlabel('Rating (1.5–5.0)', fontsize=11)
axes[1].set_ylabel('')

plt.suptitle('Category-level Price & Rating Spread (Top 10)', fontsize=15, fontweight='bold', y=1.01)
plt.tight_layout()
plt.savefig(FIGURES_PATH / '12_boxplot_category.png', bbox_inches='tight', dpi=150)
plt.show()
print('Saved: 12_boxplot_category.png')
```

### 7c. Regression Plot — Price vs Rating (Spearman r = 0.0286)

```python
sample_df = df_clean_price[['price_inr', 'rating']].dropna().sample(
    n=min(5000, len(df_clean_price)), random_state=42
)

fig, ax = plt.subplots(figsize=(10, 6))
sns.regplot(
    data=sample_df, x='price_inr', y='rating',
    scatter_kws={'alpha': 0.15, 'color': '#2196F3', 's': 20},
    line_kws={'color': '#E53935', 'linewidth': 2},
    ax=ax
)
ax.set_title(
    'Price vs Rating — Regression Analysis\n'
    'Spearman r = 0.0286  |  p < 0.001  |  Weak positive relationship',
    fontsize=13, fontweight='bold'
)
ax.set_xlabel('Price (₹) — Outliers Excluded', fontsize=11)
ax.set_ylabel('Customer Rating', fontsize=11)
ax.xaxis.set_major_formatter(mticker.FuncFormatter(lambda x, _: f'₹{int(x)}'))
ax.annotate(
    'Spearman r = 0.0286\np < 0.001\nN = {:,} (sample)'.format(len(sample_df)),
    xy=(0.72, 0.10), xycoords='axes fraction',
    fontsize=10, color='#333333',
    bbox=dict(boxstyle='round,pad=0.4', fc='#FFFDE7', alpha=0.8)
)
plt.tight_layout()
plt.savefig(FIGURES_PATH / '13_regression_price_rating.png', bbox_inches='tight', dpi=150)
plt.show()
print('Saved: 13_regression_price_rating.png')
```

### 7d. ANOVA Result — Mean Rating by City

```python
city_mean_rating = (
    df.groupby('city')['rating']
    .mean()
    .sort_values(ascending=True)
    .reset_index()
    .rename(columns={'rating': 'mean_rating'})
)

fig, ax = plt.subplots(figsize=(10, 8))
bars = sns.barplot(
    data=city_mean_rating, x='mean_rating', y='city',
    palette='YlOrRd', orient='h', ax=ax,
    errorbar=('ci', 95)
)
ax.set_title(
    'Mean Customer Rating by City\nANOVA: F = 62.97, p < 0.001 — Cities differ significantly',
    fontsize=13, fontweight='bold'
)
ax.set_xlabel('Mean Rating', fontsize=11)
ax.set_ylabel('City', fontsize=11)
ax.set_xlim(3.9, 4.7)
# Annotate each bar with its value
for bar, (_, row) in zip(ax.patches, city_mean_rating.iterrows()):
    ax.text(
        bar.get_width() + 0.005,
        bar.get_y() + bar.get_height() / 2,
        f'{row["mean_rating"]:.2f}',
        va='center', ha='left', fontsize=9
    )
plt.tight_layout()
plt.savefig(FIGURES_PATH / '14_anova_mean_rating_by_city.png', bbox_inches='tight', dpi=150)
plt.show()
print('Saved: 14_anova_mean_rating_by_city.png')
```

### 7e. ANOVA Result — Mean Rating by Top 20 Categories

```python
top20_cats = df['category'].value_counts().head(20).index.tolist()
df_top20 = df[df['category'].isin(top20_cats)]

cat_mean_rating = (
    df_top20.groupby('category')['rating']
    .mean()
    .sort_values(ascending=True)
    .reset_index()
    .rename(columns={'rating': 'mean_rating'})
)

fig, ax = plt.subplots(figsize=(10, 9))
sns.barplot(
    data=cat_mean_rating, x='mean_rating', y='category',
    palette='coolwarm', orient='h', ax=ax,
    errorbar=('ci', 95)
)
ax.set_title(
    'Mean Customer Rating by Category (Top 20)\nANOVA: F = 3.12, p < 0.001 — Categories differ significantly',
    fontsize=13, fontweight='bold'
)
ax.set_xlabel('Mean Rating', fontsize=11)
ax.set_ylabel('Category', fontsize=11)
ax.set_xlim(3.8, 4.8)
for bar, (_, row) in zip(ax.patches, cat_mean_rating.iterrows()):
    ax.text(
        bar.get_width() + 0.005,
        bar.get_y() + bar.get_height() / 2,
        f'{row["mean_rating"]:.2f}',
        va='center', ha='left', fontsize=9
    )
plt.tight_layout()
plt.savefig(FIGURES_PATH / '15_anova_mean_rating_by_category.png', bbox_inches='tight', dpi=150)
plt.show()
print('Saved: 15_anova_mean_rating_by_category.png')
```
